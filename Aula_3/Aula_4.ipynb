# Testes de Hipóteses

## Exercício a) — Empresa AAA: Impacto do Novo Produto no Faturamento

### Hipóteses

$$
H_0: \mu = \mu_0
$$

$$
H_1: \mu \neq \mu_0
$$

> **Teste bicaudal:** verifica se houve alteração no faturamento médio, sem assumir direção positiva ou negativa.

---

## Exercício b) — Indústria Farmacêutica XYZ: Novo Medicamento BBB

### Hipóteses

$$
H_0: \mu \geq 10
$$

$$
H_1: \mu < 10
$$

> **Teste unicaudal à esquerda:** verifica se o novo medicamento reduziu o tempo médio de efeito.

---

## Legenda

- $H_0$ → Hipótese nula
- $H_1$ → Hipótese alternativa
- $\mu$ → Média populacional
- $\mu_0$ → Média de referência

```python
import scipy.stats as stats
import math
 
# ============================================================
# EXERCÍCIO A — Empresa AAA: Impacto no Faturamento Médio
# ============================================================
 
print("=" * 55)
print("EXERCÍCIO A — Empresa AAA: Faturamento Médio")
print("=" * 55)
 
mu0_a   = 50000   # média anterior (hipótese nula)
xbar_a  = 54000   # média amostral
sigma_a = 8000    # desvio padrão
n_a     = 30      # tamanho da amostra
alpha_a = 0.05    # nível de significância
 
# Cálculo do Z
z_a = (xbar_a - mu0_a) / (sigma_a / math.sqrt(n_a))
 
# Valor crítico — bicaudal
z_crit_a = stats.norm.ppf(1 - alpha_a / 2)
 
# p-valor — bicaudal
p_valor_a = 2 * (1 - stats.norm.cdf(abs(z_a)))
 
print(f"\nH₀: μ = {mu0_a}  (produto não impactou o faturamento)")
print(f"H₁: μ ≠ {mu0_a}  (produto impactou o faturamento)")
print(f"\nMédia amostral  (x̄): R$ {xbar_a:,.0f}")
print(f"Desvio padrão   (σ): R$ {sigma_a:,.0f}")
print(f"Tamanho amostra (n): {n_a}")
print(f"Nível de signif.(α): {alpha_a}")
print(f"\nEstatística Z calculada : {z_a:.4f}")
print(f"Z crítico (±)           : ±{z_crit_a:.4f}")
print(f"p-valor                 : {p_valor_a:.4f}")
 
if abs(z_a) > z_crit_a:
    print("\n✅ DECISÃO: Rejeita-se H₀")
    print("→ O novo produto IMPACTOU significativamente o faturamento médio.")
else:
    print("\n❌ DECISÃO: Não se rejeita H₀")
    print("→ Não há evidência suficiente de impacto no faturamento.")
```

```python
# ============================================================
# EXERCÍCIO B — Indústria XYZ: Tempo de Efeito do Medicamento
# ============================================================
 
print("\n" + "=" * 55)
print("EXERCÍCIO B — Indústria XYZ: Tempo de Efeito (BBB)")
print("=" * 55)
 
mu0_b   = 10    # tempo médio anterior (minutos)
xbar_b  = 8.5   # média amostral
sigma_b = 3     # desvio padrão
n_b     = 25    # tamanho da amostra
alpha_b = 0.05  # nível de significância
 
# Cálculo do Z
z_b = (xbar_b - mu0_b) / (sigma_b / math.sqrt(n_b))
 
# Valor crítico — unicaudal à esquerda
z_crit_b = stats.norm.ppf(alpha_b)
 
# p-valor — unicaudal à esquerda
p_valor_b = stats.norm.cdf(z_b)
 
print(f"\nH₀: μ ≥ {mu0_b} min  (nova fórmula não reduziu o tempo)")
print(f"H₁: μ < {mu0_b} min  (nova fórmula reduziu o tempo)")
print(f"\nMédia amostral  (x̄): {xbar_b} min")
print(f"Desvio padrão   (σ): {sigma_b} min")
print(f"Tamanho amostra (n): {n_b}")
print(f"Nível de signif.(α): {alpha_b}")
print(f"\nEstatística Z calculada : {z_b:.4f}")
print(f"Z crítico               : {z_crit_b:.4f}")
print(f"p-valor                 : {p_valor_b:.4f}")
 
if z_b < z_crit_b:
    print("\n✅ DECISÃO: Rejeita-se H₀")
    print("→ A nova fórmula REDUZIU significativamente o tempo de efeito.")
    print("→ A empresa DEVE lançar o novo produto.")
else:
    print("\n❌ DECISÃO: Não se rejeita H₀")
    print("→ Não há evidência suficiente de melhora. Não lançar o produto.")
```

# Exercício 2 — Distribuição Normal

## Dados do problema

### Setembro
A temperatura em setembro segue:

$$
X_s \sim N(28, 5^2)
$$

- Média: $\mu = 28$
- Desvio padrão: $\sigma = 5$

---

### Junho
A temperatura em junho segue:

$$
X_j \sim N(32, 16)
$$

Como o segundo parâmetro é a variância:

$$
\sigma = \sqrt{16} = 4
$$

Logo:

- Média: $\mu = 32$
- Desvio padrão: $\sigma = 4$

---

# Objetivo

Determinar qual mês possui maior probabilidade de apresentar temperaturas menores que:

$$
34^\circ C
$$

---

# Setembro

Queremos calcular:

$$
P(X < 34)
$$

Padronizando:

$$
Z = \frac{34 - 28}{5}
$$

$$
Z = \frac{6}{5} = 1,2
$$

Consultando a tabela da normal padrão:

$$
P(Z < 1,2) \approx 0,8849
$$

Ou seja:

$$
P(X < 34) \approx 88,5\%
$$

---

# Junho

Queremos calcular:

$$
P(X < 34)
$$

Padronizando:

$$
Z = \frac{34 - 32}{4}
$$

$$
Z = \frac{2}{4} = 0,5
$$

Consultando a tabela da normal padrão:

$$
P(Z < 0,5) \approx 0,6915
$$

Ou seja:

$$
P(X < 34) \approx 69,15\%
$$

---

# Conclusão

A maior probabilidade de temperaturas abaixo de $34^\circ C$ ocorre em setembro.

- Setembro: $88,5\%$
- Junho: $69,15\%$

---

# Alternativa correta

$$
\boxed{\text{A}}
$$

> O melhor mês para essa pessoa viajar é setembro, pois a probabilidade da temperatura apresentar valores menores que 34 graus é de 88,5% enquanto para o mês de junho é de 69,15%.

```python
"""
Distribuição Normal - Análise de Temperatura
Problema: Qual o melhor mês para viajar (junho ou setembro)
considerando que o viajante prefere temperaturas menores que 34°C?
"""

from scipy import stats
import math

# ── Parâmetros das distribuições ────────────────────────────
# Notação: N(μ, σ²)  →  média e variância

# Setembro: N(28, 25)
mu_set  = 28
var_set = 25
sigma_set = math.sqrt(var_set)   # σ = √25 = 5

# Junho: N(32, 16)
mu_jun  = 32
var_jun = 16
sigma_jun = math.sqrt(var_jun)   # σ = √16 = 4

# Temperatura de interesse
x = 34

# ── Cálculo dos valores Z ───────────────────────────────────
# Fórmula de padronização: Z = (X - μ) / σ

z_set = (x - mu_set) / sigma_set   # Z setembro
z_jun = (x - mu_jun) / sigma_jun   # Z junho

# ── Probabilidades acumuladas P(X < 34) ─────────────────────
# Φ(z) = probabilidade acumulada da normal padrão até z

p_set = stats.norm.cdf(z_set)
p_jun = stats.norm.cdf(z_jun)

# ── Exibição dos resultados ─────────────────────────────────
print("=" * 55)
print("  DISTRIBUIÇÃO NORMAL - ANÁLISE DE TEMPERATURA")
print("=" * 55)

print(f"\n── SETEMBRO ─────────────────────────────────────────")
print(f"  Distribuição : N(μ={mu_set}, σ²={var_set})")
print(f"  Desvio padrão: σ = √{var_set} = {sigma_set}")
print(f"  Z = (X - μ) / σ = ({x} - {mu_set}) / {sigma_set} = {z_set:.2f}")
print(f"  P(X < 34) = Φ({z_set:.2f}) = {p_set:.4f}  →  {p_set*100:.2f}%")

print(f"\n── JUNHO ────────────────────────────────────────────")
print(f"  Distribuição : N(μ={mu_jun}, σ²={var_jun})")
print(f"  Desvio padrão: σ = √{var_jun} = {sigma_jun}")
print(f"  Z = (X - μ) / σ = ({x} - {mu_jun}) / {sigma_jun} = {z_jun:.2f}")
print(f"  P(X < 34) = Φ({z_jun:.2f}) = {p_jun:.4f}  →  {p_jun*100:.2f}%")

print("\n" + "=" * 55)
print("  CONCLUSÃO")
print("=" * 55)
melhor = "SETEMBRO" if p_set > p_jun else "JUNHO"
print(f"\n  Setembro → P(T < 34°C) = {p_set*100:.2f}%")
print(f"  Junho    → P(T < 34°C) = {p_jun*100:.2f}%")
print(f"\n  Melhor mês para viajar: {melhor}")
print(f"  Resposta: ALTERNATIVA A ✔")
```

# 3) Sobre Erro Padrão

---

## a) O erro padrão da média é:

### Alternativas

A – Utilizado para calcular os intervalos de confiança da média.

B – Sempre normalmente distribuídos.

C – Às vezes inferior a 0.

D – Nenhuma das alternativas.

---

## b) Cálculo do erro padrão da média

Uma amostra aleatória simples de 100 observações foi feita a partir de uma grande população.

Nesta amostra:

- Média amostral:

$$
\bar{x} = 87
$$

- Desvio padrão:

$$
s = 23
$$

- Tamanho da amostra:

$$
n = 100
$$

---

### Pergunta

Qual é o erro padrão da média?

---

### Alternativas

A – 2,30

B – 0,23

C – 8,70

D – 0,87

```python
"""
Erro Padrão da Média (EPM)
==========================
Questão A: conceito do erro padrão
Questão B: cálculo com n=100, média=87, desvio padrão=23
"""

import math

print("=" * 55)
print("  ERRO PADRÃO DA MÉDIA")
print("=" * 55)

# ── Questão A ───────────────────────────────────────────────
print("\n── QUESTÃO A ────────────────────────────────────────")
print("""
  O erro padrão da média (EPM) mede a variabilidade
  da média amostral em relação à média populacional.

  Sua principal aplicação é:
  → Construir INTERVALOS DE CONFIANÇA para a média.

  Resposta: ALTERNATIVA A ✔

  Por que as outras estão erradas?
  B) NÃO é sempre normalmente distribuído (depende do n).
  C) NUNCA é negativo, pois EPM = s / √n >= 0.
""")

# ── Questão B ───────────────────────────────────────────────
print("── QUESTÃO B ────────────────────────────────────────")

n     = 100   # tamanho da amostra
media = 87    # média amostral (x̄)
s     = 23    # desvio padrão amostral

# Fórmula do Erro Padrão da Média:
#
#         s
# EPM = -----
#        √ n
#

epm = s / math.sqrt(n)

print(f"\n  Dados:")
print(f"    n  = {n}  (tamanho da amostra)")
print(f"    x̄  = {media}  (média amostral)")
print(f"    s  = {s}  (desvio padrão amostral)")

print(f"\n  Fórmula:")
print(f"    EPM = s / √n")
print(f"    EPM = {s} / √{n}")
print(f"    EPM = {s} / {math.sqrt(n):.1f}")
print(f"    EPM = {epm:.2f}")

print(f"\n  Resposta: ALTERNATIVA A ✔  →  EPM = {epm:.2f}")
print("=" * 55)
```

# 4) Distribuição Normal — Tempo de Exame Vestibular

O tempo gasto no exame vestibular de uma universidade possui distribuição Normal com:

- Média:

$$
\mu = 120 \text{ min}
$$

- Desvio padrão:

$$
\sigma = 15 \text{ min}
$$

Logo:

$$
X \sim N(120, 15^2)
$$

---

# a) Probabilidade de terminar o exame antes de 100 minutos

Deseja-se calcular:

$$
P(X < 100)
$$

Padronizando a variável:

$$
Z = \frac{100 - 120}{15}
$$

---

# b) Probabilidade de terminar o exame depois de 150 minutos

Deseja-se calcular:

$$
P(X > 150)
$$

Padronizando a variável:

$$
Z = \frac{150 - 120}{15}
$$

---

# c) Tempo necessário para que 95% dos alunos terminem a prova

Deseja-se encontrar o valor de:

$$
x
$$

tal que:

$$
P(X < x) = 0,95
$$

Utilizando a fórmula da normal padronizada:

$$
Z = \frac{x - \mu}{\sigma}
$$

Substituindo os valores conhecidos:

$$
Z = \frac{x - 120}{15}
$$

Encontrar o valor de:

$$
x
$$

correspondente ao percentil de 95%.

```python
"""
Distribuição Normal - Tempo de Vestibular
==========================================
Distribuição: N(μ=120 min, σ=15 min)

a) P(X < 100)  → probabilidade de terminar antes de 100 min
b) P(X > 150)  → probabilidade de terminar depois de 150 min
c) Tempo t tal que 95% dos alunos terminem no prazo
"""

from scipy import stats
import math

# ── Parâmetros da distribuição ───────────────────────────────
mu    = 120   # média em minutos
sigma = 15    # desvio padrão em minutos

print("=" * 58)
print("  DISTRIBUIÇÃO NORMAL - TEMPO DE VESTIBULAR")
print(f"  N(μ={mu} min, σ={sigma} min)")
print("=" * 58)

# ── a) P(X < 100 min) ───────────────────────────────────────
xa = 100
za = (xa - mu) / sigma      # Z = (X - μ) / σ
pa = stats.norm.cdf(za)     # P(X < x) = Φ(Z)

print(f"\n── a) P(X < {xa} min) ──────────────────────────────────")
print(f"  Código usado:")
print(f"    za = ({xa} - {mu}) / {sigma}          # Z = (X - μ) / σ")
print(f"    pa = stats.norm.cdf(za)        # P(X < x) = Φ(Z)")
print(f"  Resultado:")
print(f"    za = {za:.4f}")
print(f"    pa = {pa:.4f}  →  {pa*100:.2f}%")

# ── b) P(X > 150 min) ───────────────────────────────────────
xb = 150
zb = (xb - mu) / sigma              # Z = (X - μ) / σ
pb = 1 - stats.norm.cdf(zb)         # P(X > x) = 1 - Φ(Z)

print(f"\n── b) P(X > {xb} min) ──────────────────────────────────")
print(f"  Código usado:")
print(f"    zb = ({xb} - {mu}) / {sigma}          # Z = (X - μ) / σ")
print(f"    pb = 1 - stats.norm.cdf(zb)    # P(X > x) = 1 - Φ(Z)")
print(f"  Resultado:")
print(f"    zb = {zb:.4f}")
print(f"    pb = {pb:.4f}  →  {pb*100:.2f}%")

# ── c) Tempo t para 95% dos alunos terminarem ───────────────
p_alvo = 0.95
zc = stats.norm.ppf(p_alvo)   # Z crítico = Φ⁻¹(p)
tc = mu + zc * sigma           # t = μ + Z × σ

print(f"\n── c) Tempo t tal que P(X < t) = 95% ──────────────────")
print(f"  Código usado:")
print(f"    zc = stats.norm.ppf({p_alvo})       # Z crítico = Φ⁻¹(p)")
print(f"    tc = {mu} + zc * {sigma}             # t = μ + Z × σ")
print(f"  Resultado:")
print(f"    zc = {zc:.4f}")
print(f"    tc = {tc:.2f} min  →  aproximadamente {math.ceil(tc)} min")

print("\n" + "=" * 58)
print("  RESUMO")
print("=" * 58)
print(f"  a) P(X < 100 min) = {pa*100:.2f}%")
print(f"  b) P(X > 150 min) = {pb*100:.2f}%")
print(f"  c) Tempo para 95% terminarem ≈ {tc:.2f} min")
print("=" * 58)
```

US\$ 50

```python
"""
Distribuição Normal - Tempo de Vestibular
==========================================
Distribuição: N(μ=120 min, σ=15 min)

a) P(X < 100)  → probabilidade de terminar antes de 100 min
b) P(X > 150)  → probabilidade de terminar depois de 150 min
c) Tempo t tal que 95% dos alunos terminem no prazo
"""

from scipy import stats
import math

# ── Parâmetros da distribuição ───────────────────────────────
mu    = 120   # média em minutos
sigma = 15    # desvio padrão em minutos

print("=" * 58)
print("  DISTRIBUIÇÃO NORMAL - TEMPO DE VESTIBULAR")
print(f"  N(μ={mu} min, σ={sigma} min)")
print("=" * 58)

# ── a) P(X < 100 min) ───────────────────────────────────────
xa = 100
za = (xa - mu) / sigma      # Z = (X - μ) / σ
pa = stats.norm.cdf(za)     # P(X < x) = Φ(Z)

print(f"\n── a) P(X < {xa} min) ──────────────────────────────────")
print(f"  Código usado:")
print(f"    za = ({xa} - {mu}) / {sigma}          # Z = (X - μ) / σ")
print(f"    pa = stats.norm.cdf(za)        # P(X < x) = Φ(Z)")
print(f"  Resultado:")
print(f"    za = {za:.4f}")
print(f"    pa = {pa:.4f}  →  {pa*100:.2f}%")

# ── b) P(X > 150 min) ───────────────────────────────────────
xb = 150
zb = (xb - mu) / sigma              # Z = (X - μ) / σ
pb = 1 - stats.norm.cdf(zb)         # P(X > x) = 1 - Φ(Z)

print(f"\n── b) P(X > {xb} min) ──────────────────────────────────")
print(f"  Código usado:")
print(f"    zb = ({xb} - {mu}) / {sigma}          # Z = (X - μ) / σ")
print(f"    pb = 1 - stats.norm.cdf(zb)    # P(X > x) = 1 - Φ(Z)")
print(f"  Resultado:")
print(f"    zb = {zb:.4f}")
print(f"    pb = {pb:.4f}  →  {pb*100:.2f}%")

# ── c) Tempo t para 95% dos alunos terminarem ───────────────
p_alvo = 0.95
zc = stats.norm.ppf(p_alvo)   # Z crítico = Φ⁻¹(p)
tc = mu + zc * sigma           # t = μ + Z × σ

print(f"\n── c) Tempo t tal que P(X < t) = 95% ──────────────────")
print(f"  Código usado:")
print(f"    zc = stats.norm.ppf({p_alvo})       # Z crítico = Φ⁻¹(p)")
print(f"    tc = {mu} + zc * {sigma}             # t = μ + Z × σ")
print(f"  Resultado:")
print(f"    zc = {zc:.4f}")
print(f"    tc = {tc:.2f} min  →  aproximadamente {math.ceil(tc)} min")

print("\n" + "=" * 58)
print("  RESUMO")
print("=" * 58)
print(f"  a) P(X < 100 min) = {pa*100:.2f}%")
print(f"  b) P(X > 150 min) = {pb*100:.2f}%")
print(f"  c) Tempo para 95% terminarem ≈ {tc:.2f} min")
print("=" * 58)
```
